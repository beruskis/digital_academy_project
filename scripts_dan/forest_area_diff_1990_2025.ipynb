```python
import pandas as pd
from pathlib import Path

SRC  = Path('../data_raw/MAIN_1990-2025forest-area-as-share-of-land-area.csv')
OUT  = Path('../data_dan/MAIN_1990-2025forest_difference.csv')
VALUE_COL = 'Share of land covered by forest'
ANNOT_COL = 'Share of land covered by forest (Annotations)'

df = pd.read_csv(SRC)
df = df[df['Year'] != 'diff_2025_minus_1990'].copy()

y1990 = df[df['Year'] == 1990].set_index('Entity')[['Code', VALUE_COL]]
y2025 = df[df['Year'] == 2025].set_index('Entity')[[VALUE_COL]]
common = y1990.index.intersection(y2025.index)

new_rows = pd.DataFrame({
    'Entity': common,
    'Code': y1990.loc[common, 'Code'].values,
    'Year': 'diff_2025_minus_1990',
    VALUE_COL: (y2025.loc[common, VALUE_COL] - y1990.loc[common, VALUE_COL]).values,
    ANNOT_COL: ''
})

new_rows.to_csv(OUT, index=False)
print(f'Done. Saved {len(new_rows)} diff rows to {OUT}')
```
