## Problem
The file has issues that must be fixed before use:
1. **4 metadata rows** at the top — the real header is on row 5, so rows 1-4 must be skipped.
2. **17 country names contain commas** (e.g. `Congo, Dem. Rep.`, `Korea, Rep.`, `Bahamas, The`) — these break any parser that does not respect CSV quoting, causing column misalignment.
3. **Trailing comma on every row** creates a spurious empty `Unnamed` column.
4. **Empty 2025 column** — the file is titled 1960-2024 but has a blank 2025 placeholder.

Issues 3 and 4 are fixed by dropping all fully-empty columns after loading.

```python
import pandas as pd
from pathlib import Path

SRC = Path('../data_raw/1960-2024_GDP_USD.csv')
DST = Path('../data_raw/1960-2024_GDP_USD_clean.csv')

# skiprows=4 skips the metadata block; pandas respects CSV quoting for names with commas
df = pd.read_csv(SRC, skiprows=4)

# Drop fully-empty columns (trailing comma artefact + empty 2025 placeholder)
df = df.dropna(axis=1, how='all')

# Replace commas in country names with ' -'
df['Country Name'] = df['Country Name'].str.replace(', ', ' - ', regex=False)

df.to_csv(DST, index=False)
print(f'Saved to {DST}  ({len(df)} rows, {len(df.columns)} columns)')
df.head(3)
```
