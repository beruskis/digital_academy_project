```python
import pandas as pd
```

```python
df_forest = pd.read_csv("MAIN_1990-2025forest-area-as-share-of-land-area.csv")
```

```python
display(df_forest)
```

```python
# Get 1990 and 2025 values for each country
y1990 = df_forest[df_forest['Year'] == 1990][['Code', 'Share of land covered by forest']].rename(columns={'Share of land covered by forest': 'forest_1990'})
y2025 = df_forest[df_forest['Year'] == 2025][['Code', 'Share of land covered by forest']].rename(columns={'Share of land covered by forest': 'forest_2025'})

# Merge them together
diff_df_forest = y1990.merge(y2025, on='Code')

# Calculate the difference
diff_df_forest['forest_change_1990_2025'] = diff_df_forest['forest_2025'] - diff_df_forest['forest_1990']

print(diff_df_forest.head(10))
```

```python
# Merge the new columns back into the original dataframe
diff_df_forest = df_forest.merge(
    diff_df_forest[['Code', 'forest_1990', 'forest_2025', 'forest_change_1990_2025']],
    on='Code',
    how='left'        # keeps ALL original rows, even if no match
)

# Check it looks right
print(diff_df_forest.head(10))
print(diff_df_forest.shape)        # should have same rows as original + 3 new columns
print(diff_df_forest.columns.tolist())

# Save to a new file in processed folder
diff_df_forest.to_csv('/Users/beruska/Desktop/Project/CODE/digital_academy_project/new.csv', index=False)
```
