```python
import pandas as pd
data=pd.read_csv("1960-2024_GDP_USD_clean.csv")
data.head()
```

```python
df_long=data.melt(id_vars=["Country Name", "Country Code", "Indicator Name", "Indicator Code"],
                  var_name="Year",
                  value_name="GDP")
df_long = df_long.sort_values(by=['Country Name', 'Year'])
df_long.head(100)
```

```python
df_long.to_csv('GDP_pivot.csv', index=False, encoding='utf-8-sig')
```
